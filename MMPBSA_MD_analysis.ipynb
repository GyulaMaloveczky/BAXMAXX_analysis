```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
from pathlib import Path

BASE = Path("/FastHome/gyula/designs/place_holder/DATA/"
            "gy_2nd_iter_try_29_balm_deployed_weight_10_length_15/"
            "md_sim_analysis/mmpbsa/jobs/")
FNAME = "test_results_energy.csv"
COL = "TOTAL"
FPS = 40          # frames per ns


def load(path, col=COL, max_skip=5):
    """Read csv, scanning the first few rows for the one that is the header."""
    for skip in range(max_skip):
        try:
            df = pd.read_csv(path, skiprows=1616)
        except (pd.errors.EmptyDataError, pd.errors.ParserError):
            continue
        df.columns = [str(c).strip() for c in df.columns]
        if col in df.columns:
            return df.dropna(subset=[col]).reset_index(drop=True)
    return None


dataframes = {}
for path in sorted(BASE.rglob(FNAME)):
    df = load(path)
    if df is None or df.empty:
        print(f"Warning: '{COL}' not found (or no data) in {path}. Skipping.")
        continue
    df['cumulative_avg'] = df[COL].cumsum() / np.arange(1, len(df) + 1)
    dataframes[path.parent.name] = df

if not dataframes:
    raise RuntimeError(f"No usable '{FNAME}' files found under {BASE}.")

print(f"Loaded {len(dataframes)} jobs: {list(dataframes)[:5]}...")
```

```python

if not dataframes:
    raise RuntimeError(f"No usable '{FNAME}' files found under {BASE}.")

print(f"Loaded {len(dataframes)} jobs: {list(dataframes)}")

def cum_avg(df, start):
    """Cumulative average recomputed from frame `start` onward."""
    if len(df) <= start:
        return None
    sub = df.iloc[start:].reset_index(drop=True)
    return sub[COL].cumsum() / np.arange(1, len(sub) + 1)


keys = list(dataframes)
fig, axes = plt.subplots(nrows=len(keys), ncols=1,
                         figsize=(10, 6 * len(keys)), squeeze=False)
axes = axes.ravel()

for ax, key in zip(axes, keys):
    df = dataframes[key].reset_index(drop=True)
    t = np.arange(len(df)) / FPS          # x-axis in ns

    ax.plot(t, df['cumulative_avg'], label=f"{key} (uncut)",
            color='blue', marker='o', markevery=50, linestyle='-')

    for start, colour, mark, ls in ((40, 'green', '.', '--'),
                                    (80, 'red', 'x', '-.')):
        ca = cum_avg(df, start)
        if ca is None:
            print(f"Warning: {key} shorter than {start} frames; skipping cut.")
            continue
        ax.plot(t[start:], ca, label=f"{key} (cut at {start/FPS:g} ns)",
                color=colour, marker=mark, markevery=50, linestyle=ls)

    ax.set_title(f"Cumulative average energy delta: {key}")
    ax.set_xlabel("simulation length (ns)")
    ax.set_ylabel("MMPBSA (kJ/mol)")
    ax.set_ylim(-90, -50)
    ax.set_xlim(0, len(df) / FPS)
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

fig.tight_layout()
plt.show()
```
