```python
import pyrosetta
from pyrosetta.rosetta.protocols.analysis import InterfaceAnalyzerMover

pyrosetta.init("-ex1 -ex2 -use_input_sc -ignore_unrecognized_res")

pose = pyrosetta.pose_from_pdb("complex.pdb")
# Identify the jump index connecting the two chains
jump_num = 1

scorefxn = pyrosetta.get_fa_scorefxn()
task_pack = pyrosetta.standard_packer_task(pose)
task_pack.restrict_to_repacking()

pack_mover = pyrosetta.rosetta.protocols.minimization_packing.PackRotamersMover(scorefxn, task_pack)
pack_mover.apply(pose)

# 'chain_id' specifies the jump across which to separate the partners
iam = InterfaceAnalyzerMover(jump_num, scorefxn)
iam.set_compute_packstat(True)
iam.set_pack_separated(True) # Re-packs side chains after separation for accuracy
iam.apply(pose)

# Extraction of thermodynamic metrics
interface_energy = iam.get_interface_dG()
interface_delta_h_bond = iam.get_interface_delta_hbond()
```
