```python
import pickle
import matplotlib.pyplot as plt
import torch
import json
# "/home/kunzj/BindCraft_uva_internship/surr_model/training_results_jax/results_mmpbsa_three_way.pkl"
# '/home/kunzj/BindCraft_uva_internship/surr_model/training_results_jax/results_ppi_three_way.pkl'
with open("/home/kunzj/BindCraft_uva_internship/surr_model/pytorch_implementation/models/first_training/model_performance.json", 'rb') as f:
    data = pickle.load(f)

# Plot training and validation losses
plt.figure(figsize=(10, 6))
plt.plot(data['train_loss_epoch'], label='Train Loss')
# Assuming val_loss_epoch is your original list
cleaned_val_losses = [loss.detach().cpu().item() for loss in data["val_loss_epoch"]]
plt.plot(cleaned_val_losses, label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.title('BALM Model Training Progress')
plt.grid(True)
plt.show()

# Find best validation loss
best_epoch = data['val_losses'].index(min(data['val_losses']))
print(f"Best validation loss: {min(data['val_losses']):.6f} at epoch {best_epoch}")
```
